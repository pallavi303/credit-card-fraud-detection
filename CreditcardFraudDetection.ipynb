```python
#train test & split
from sklearn.model_selection import train_test_split
x=df.drop('Class',axis=1)
y=df['Class']
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30,random_state=42)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
#decision tree
from sklearn.tree import DecisionTreeClassifier
dtree=DecisionTreeClassifier()
dtree.fit(x_train,y_train)
```

```python
#prediction
predictions=dtree.predict(x_test)
```

```python
predictions
```

```python
#confusion matrix
from sklearn.metrics import confusion_matrix,classification_report
print(confusion_matrix(y_test,predictions))
```

```python
#classification Report
print(classification_report(y_test,predictions))
```

```python
#visualize the data using tree
from sklearn import tree
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))

tree.plot_tree(
    dtree,
    filled=True,
    rounded=True,
    max_depth=2,
    fontsize=10
)

plt.show()
```

```python
df['Class'].value_counts().plot(kind='bar')
plt.title('Normal vs Fraud')
plt.show()
```

```python
#standerdize the variable
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x=pd.DataFrame(scaler.fit_transform(df.drop('Class',axis=1)))
y=df['Class']
```

```python
x.head()
```

```python
#train test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30,random_state=0)
```

```python
#using knn
from sklearn.neighbors import KNeighborsClassifier
knn=KNeighborsClassifier(n_neighbors=1)
knn.fit(x_train,y_train)
```

```python
#prediction
pred=knn.predict(x_test)
pred
```

```python
#prediction and evaluation
from sklearn.metrics import classification_report,confusion_matrix
print(confusion_matrix(y_test,pred))
```

```python
#visualizing
sns.heatmap(confusion_matrix(y_test,pred),annot=True)
plt.show()
```

```python
#display data after apply knn
print(classification_report(y_test,pred))
```

```python
import pandas as pd
from IPython.display import display, HTML, clear_output
import ipywidgets as widgets

# ------------------------------------------------------------
# BEST ALGORITHM
# ------------------------------------------------------------

best_algorithm = "Decision Tree"

# ------------------------------------------------------------
# FEATURE COLUMNS
# ------------------------------------------------------------

feature_columns = ["Time"]

for i in range(1, 29):
    feature_columns.append(f"V{i}")

feature_columns.append("Amount")

# ------------------------------------------------------------
# HEADER
# ------------------------------------------------------------

display(
    HTML(
        f"""
        <div style="
            text-align:center;
            background:#202124;
            color:white;
            padding:25px;
            border-radius:15px;
        ">
        <h1>💳 Credit Card Fraud Detection System</h1>

        <h3>
        Best Algorithm:
        <span style="color:#81c995;">
        {best_algorithm}
        </span>
        </h3>

        <p>
        Load a sample transaction and check the prediction
        </p>
        </div>
        """
    )
)

# ------------------------------------------------------------
# INPUT BOXES
# ------------------------------------------------------------

time_input = widgets.FloatText(
    description="Time:",
    value=0,
    layout=widgets.Layout(width="400px")
)

amount_input = widgets.FloatText(
    description="Amount:",
    value=0,
    layout=widgets.Layout(width="400px")
)

# ------------------------------------------------------------
# V1 - V28
# ------------------------------------------------------------

v_inputs = []

for i in range(1, 29):

    widget = widgets.FloatText(
        description=f"V{i}:",
        value=0,
        layout=widgets.Layout(width="250px")
    )

    v_inputs.append(widget)

# ------------------------------------------------------------
# DISPLAY INPUTS
# ------------------------------------------------------------

display(HTML("<h2>💳 Transaction Details</h2>"))

display(time_input)
display(amount_input)

display(HTML("<h3>🔢 Transaction Features</h3>"))

left_column = widgets.VBox(v_inputs[:14])
right_column = widgets.VBox(v_inputs[14:])

display(
    widgets.HBox(
        [left_column, right_column]
    )
)

# ------------------------------------------------------------
# LOAD SAMPLE BUTTON
# ------------------------------------------------------------

sample_button = widgets.Button(
    description="🎲 Load Sample Transaction",
    button_style="info",
    layout=widgets.Layout(
        width="250px",
        height="45px"
    )
)

# ------------------------------------------------------------
# CHECK BUTTON
# ------------------------------------------------------------

check_button = widgets.Button(
    description="🔍 Check Transaction",
    button_style="success",
    layout=widgets.Layout(
        width="220px",
        height="45px"
    )
)

output = widgets.Output()

# ------------------------------------------------------------
# LOAD SAMPLE FUNCTION
# ------------------------------------------------------------

def load_sample(b):

    row = df.iloc[0]

    time_input.value = float(row["Time"])
    amount_input.value = float(row["Amount"])

    for i in range(1, 29):
        v_inputs[i-1].value = float(row[f"V{i}"])

    with output:
        clear_output()

        display(
            HTML(
                "<h3>✅ Sample transaction loaded!</h3>"
            )
        )

# ------------------------------------------------------------
# PREDICTION FUNCTION
# ------------------------------------------------------------

def predict_fraud(b):

    with output:

        clear_output()

        values = {
            "Time": time_input.value
        }

        for i, widget in enumerate(v_inputs, 1):
            values[f"V{i}"] = widget.value

        values["Amount"] = amount_input.value

        input_data = pd.DataFrame([values])

        input_data = input_data[feature_columns]

        # Decision Tree
        if best_algorithm == "Decision Tree":

            prediction = dtree.predict(input_data)[0]

            probability = dtree.predict_proba(
                input_data
            )[0][1]

        # KNN
        else:

            input_scaled = scaler.transform(
                input_data
            )

            prediction = knn.predict(
                input_scaled
            )[0]

            probability = knn.predict_proba(
                input_scaled
            )[0][1]

        # ----------------------------------------------------
        # RESULT
        # ----------------------------------------------------

        if prediction == 1:

            display(
                HTML(
                    f"""
                    <div style="
                        background:#ffebee;
                        padding:20px;
                        border-radius:12px;
                        border:2px solid #ef5350;
                        text-align:center;
                    ">

                    <h2 style="color:#c62828;">
                    ⚠️ FRAUDULENT TRANSACTION
                    </h2>

                    <b>Fraud Probability:</b>
                    {probability * 100:.2f}%

                    <br><br>

                    <b>Algorithm:</b>
                    {best_algorithm}

                    </div>
                    """
                )
            )

        else:

            display(
                HTML(
                    f"""
                    <div style="
                        background:#e8f5e9;
                        padding:20px;
                        border-radius:12px;
                        border:2px solid #66bb6a;
                        text-align:center;
                    ">

                    <h2 style="color:#2e7d32;">
                    ✅ NORMAL TRANSACTION
                    </h2>

                    <b>Fraud Probability:</b>
                    {probability * 100:.2f}%

                    <br><br>

                    <b>Algorithm:</b>
                    {best_algorithm}

                    </div>
                    """
                )
            )

# ------------------------------------------------------------
# CONNECT BUTTONS
# ------------------------------------------------------------

sample_button.on_click(load_sample)
check_button.on_click(predict_fraud)

# ------------------------------------------------------------
# DISPLAY BUTTONS
# ------------------------------------------------------------

display(HTML("<br>"))

display(sample_button)

display(HTML("<br>"))

display(check_button)

display(output)
```
